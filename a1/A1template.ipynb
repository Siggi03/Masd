### Exercise 4

The precision will be down to $\frac{1}{2^{30}}$, with a maximum of 1000 iterations

```python
# Function for finding roots of "fun" in interval [a,b]
# Algorithm will assumme that supplied function is continuous
# Returns x value for root
# Returns None if invalid input (fun(a)*fun(b) >=0)
def BisectionAlgorithm(fun, a, b):
    precision = 1/(2**30)
    # Check for invalid intervals
    if (fun(a)*fun(b) >= 0):
        print("Invalid interval")
        return None

    #Maximum of 1000 iterations
    for i in range(1000):
        c = a/2 + b/2
        if (fun(c) == 0 or (b-a)/2 < precision): #Root found, or approximation reached
            return c
        elif (fun(a) * fun(c) < 0):
            b = c # Search left
        else:
            a = c # Search right
```

```python
# The function we wish to find roots of:
def function(x):
    return (13/99)*(x**3)+(x**2)/99-(269/99)*x-(47/33)

# Plot for helping to visualize the roots location
x = np.arange(-7.5, 7, 0.1)
plt.axis((-6, 6, -8, 5))
plt.grid(True)
plt.plot(x, function(x))

print(f"Root found at x= {BisectionAlgorithm(function, -6, -2)}")
print(f"Root found at x= {BisectionAlgorithm(function, -2, 2)}")
print(f"Root found at x= {BisectionAlgorithm(function, -5, 5)}")
```
