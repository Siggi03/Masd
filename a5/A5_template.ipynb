```python
# Load packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the excel file
df = pd.read_excel("MASD2024_data.xlsx")

# Extract the relevant data
nervousness_col = "The thought of being enrolled in a mathematics/statistics course makes me nervous."


school_col = "What type of high-school did you attend?"

# Remove incomplete data
data = df[[nervousness_col, school_col]].dropna()

# Setup high and low nervousness
data = data.copy()

data["High_nervousness"] = data[nervousness_col] >= 5
data["Low_nervousness"] = data[nervousness_col] < 5

data["HTX"] = data[school_col] == "HTX"
data["STX"] = data[school_col] == "STX"

print(f"Amount of HTX students: {data["HTX"].sum()}")
print(f"Amount of STX students: {data["STX"].sum()}")

print(f"Total amount of valid student surveys: {len(data)}\n")

print(f"Number of students with high nervousness: {data["High_nervousness"].sum()}")
```

# 3.d: Plot the given CDF for expected grades. Identify the region(s) where the probability mass is concentrated.

```python
def F(x):
    if x < -3:
        return 0
    elif x < 2:
        return 0.0588 * (x + 3)
    elif x < 7:
        return 0.294 + 0.1176 * (x - 2)
    elif x < 12:
        return 0.882 + 0.0235 * (x - 7)
    else:
        return 1

x = np.linspace(-5, 14, 500)
y = [F(xi) for xi in x]

plt.plot(x, y)
plt.xlabel("Expected grade x")
plt.ylabel("F(x)")
plt.title("CDF of expected grades")
plt.grid()
plt.show()

# Probability mass in each interval
print("P(-3 <= X < 2) =", F(2) - F(-3))
print("P(2 <= X < 7)  =", F(7) - F(2))
print("P(7 <= X < 12) =", F(12) - F(7))
```
